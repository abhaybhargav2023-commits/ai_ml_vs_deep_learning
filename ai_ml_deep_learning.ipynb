# AI vs ML vs Deep Learning

## Introduction
Artificial Intelligence (AI), Machine Learning (ML), and Deep Learning (DL)
are related fields used to develop intelligent computer systems.

This project compares their definitions, differences, relationships,
and real-world applications.

## Objectives
- Understand the relationship between AI, ML, and Deep Learning.
- Compare their approaches and data requirements.
- Explore practical applications.
- Visualize their conceptual relationship.

## Tools
- Python
- Jupyter Notebook
- Matplotlib
- Markdown

## 1. Artificial Intelligence (AI)

AI is the broader field of creating systems that perform tasks
normally associated with human intelligence.

Examples:
- Virtual assistants
- Rule-based expert systems
- Robotics
- Intelligent chatbots

## 2. Machine Learning (ML)

Machine Learning is a subset of AI in which systems learn patterns
from data to make predictions or decisions.

Examples:
- Email spam detection
- Recommendation systems
- Fraud detection
- House-price prediction

## 3. Deep Learning (DL)

Deep Learning is a subset of Machine Learning that uses neural
networks with multiple layers to learn complex patterns.

Examples:
- Image recognition
- Speech recognition
- Language translation
- Generative AI applications

## Comparison of AI, ML, and Deep Learning

| Feature | AI | Machine Learning | Deep Learning |
|---|---|---|---|
| Scope | Broadest field | Subset of AI | Subset of ML |
| Main approach | Rules, logic, search, or learning | Learns patterns from data | Uses multi-layer neural networks |
| Data requirement | Depends on the approach | Usually uses training data | Often benefits from substantial data |
| Feature engineering | Depends on the method | May require manual features | Can learn representations automatically |
| Example | Expert system | Spam detection | Image recognition |

## Relationship Between AI, ML, and Deep Learning

AI is the broad field of creating intelligent systems.

Machine Learning is one approach within AI, and Deep Learning is
one approach within Machine Learning.

The conceptual relationship is:

Deep Learning ⊂ Machine Learning ⊂ Artificial Intelligence

Not every AI system uses Machine Learning, and not every
Machine Learning system uses Deep Learning.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

fig, ax = plt.subplots(figsize=(8, 6))

# Outer circle: Artificial Intelligence
ai_circle = Circle(
    (0, 0), 3.0,
    facecolor="#DCEBFA",
    edgecolor="#2864A5",
    linewidth=2
)

# Middle circle: Machine Learning
ml_circle = Circle(
    (0, -0.35), 2.0,
    facecolor="#B9E8D0",
    edgecolor="#23845B",
    linewidth=2
)

# Inner circle: Deep Learning
dl_circle = Circle(
    (0, -0.75), 1.05,
    facecolor="#D9C6FA",
    edgecolor="#7542B5",
    linewidth=2
)

ax.add_patch(ai_circle)
ax.add_patch(ml_circle)
ax.add_patch(dl_circle)

ax.text(
    0, 2.35, "Artificial Intelligence (AI)",
    ha="center", fontsize=13, weight="bold", color="#174A80"
)

ax.text(
    0, 0.55, "Machine Learning (ML)",
    ha="center", fontsize=12, weight="bold", color="#155D43"
)

ax.text(
    0, -0.85, "Deep Learning\n(DL)",
    ha="center", va="center", fontsize=12, weight="bold",
    color="#4D267F"
)

ax.set_title(
    "Relationship Between AI, ML and Deep Learning",
    fontsize=14, pad=15
)

ax.set_xlim(-3.3, 3.3)
ax.set_ylim(-3.2, 3.2)
ax.set_aspect("equal")
ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
from IPython.display import display, Markdown

comparison = """
| Feature | AI | ML | Deep Learning |
|---|---|---|---|
| Scope | Broadest | Subset of AI | Subset of ML |
| Approach | Rules, logic, search, learning | Learns from data | Multi-layer neural networks |
| Example | Expert system | Spam detection | Image recognition |
"""

display(Markdown(comparison))
```
